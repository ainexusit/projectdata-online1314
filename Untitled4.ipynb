```python
def file_read(filename):
    fp = open(filename)
    data = fp.read()
    return data

def filter_email_ids():
    pass

def filter_mobile_nos(fdata):
    valid_mobile_nos = []
    invalid_mobile_nos = []
    for word in fdata.split():
        if word.isnumeric():
            if len(word)==10:
                if word.startswith('9') or word.startswith('8'):
                    valid_mobile_nos.append(word)
                else:
                    invalid_mobile_nos.append(word)
    return valid_mobile_nos,invalid_mobile_nos
                

def main():
    filename = 'sample.txt'
    fdata = file_read(filename)
    valid,invalid = filter_mobile_nos(fdata)
    print(valid)
    print(invalid)
    

if __name__=='__main__':
    main()
```
